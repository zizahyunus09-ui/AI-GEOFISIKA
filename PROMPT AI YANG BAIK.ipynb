Dalam Diagram Wadati, gunakan konsep dasar waktu tiba gelombang seismik:

- $t_P$ = waktu tiba gelombang P
- $t_S$ = waktu tiba gelombang S
- $t_0$ = waktu asal gempa
- $V_P$ = kecepatan gelombang P
- $V_S$ = kecepatan gelombang S

Salah satu hubungan dasar yang digunakan adalah:

$$
t_S-t_P =
\frac{V_P}{V_S}(t_P-t_0)
$$

Hubungan tersebut digunakan untuk membentuk Diagram Wadati dan
memperoleh parameter yang berkaitan dengan $V_P/V_S$ serta $t_0$.
