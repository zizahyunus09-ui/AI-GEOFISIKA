## 3. Praktik Recipe: Iterative Prompt Refinement
**Kaitan buku:** Chapter 2.4.7 — siklus *Initial Prompt → Evaluate Output → Identify Gaps → Follow-up Prompt → Verification*.

Latihan berikut menunjukkan bagaimana prompt dapat ditingkatkan secara bertahap, bukan mengubah semua instruksi sekaligus.

```python
prompt_awal = "Hitung selisih waktu gelombang P dan S."

evaluasi_awal = [
    "Belum menyebut format waktu yang digunakan.",
    "Belum menentukan satuan hasil.",
    "Belum menjelaskan pemeriksaan jika waktu S lebih awal dari P."
]

prompt_lanjutan = """
Diberikan waktu tiba gelombang P dan S dalam detik relatif terhadap awal rekaman.
Hitung Δt = Ts - Tp untuk setiap stasiun dan tampilkan hasil dalam detik dengan 2 angka desimal.
Jangan mengubah nilai input. Jika Ts < Tp atau salah satu nilai hilang, tandai baris tersebut
sebagai perlu diperiksa dan jangan menafsirkannya sebagai hasil valid.
"""

print("Prompt awal:", prompt_awal)
print("\nCelah yang ditemukan:")
for item in evaluasi_awal:
    print("-", item)
print("\nPrompt setelah refinement:\n", prompt_lanjutan)
```

**Verifikasi refinement:** perubahan yang dibuat menjawab celah yang ditemukan: satuan dan format, aturan validasi, serta perlakuan terhadap data yang bermasalah. Dalam penggunaan nyata, jawaban AI masih perlu dibandingkan dengan input dan metode yang dinyatakan.

# 6. Recipe Mandiri — QA/QC sederhana data waktu tiba gempa
Recipe ini dibuat sendiri untuk tugas, dengan struktur disesuaikan pada Chapter 4.2–4.5 buku.

## Identitas
- **Recipe ID:** `R-GEO-001`
- **Judul:** Pemeriksaan kualitas data waktu tiba gelombang P dan S
- **Prompt ID:** `P-GEO-001`
- **Domain:** Seismologi / Geofisika
- **Status:** Praktik ilustratif dengan data sintetis

## ROLE
Bertindak sebagai asisten pemrosesan data seismologi yang berhati-hati terhadap satuan, nilai hilang, dan asumsi.

## GOAL
Melakukan pemeriksaan kualitas (QA/QC) dasar pada data waktu tiba gelombang P dan S, lalu menghitung selisih waktu S–P untuk baris yang memiliki input lengkap.

## CONTEXT
Hasil ini ditujukan untuk mahasiswa Geofisika sebagai pemeriksaan awal sebelum analisis seismologi lanjutan. QC sederhana tidak membuktikan bahwa pick waktu tiba sudah benar.

## INPUT
Tabel yang memuat nama stasiun, `Tp_s` (waktu tiba P dalam detik), dan `Ts_s` (waktu tiba S dalam detik). Data dalam notebook ini sintetis.

## TASK
Periksa kelengkapan data, hitung selisih waktu S–P, dan beri status QA/QC pada setiap baris.

## SCOPE
1. Memeriksa apakah `Tp_s` dan `Ts_s` tersedia.
2. Menghitung \\(\\Delta t = T_S - T_P\\)\\ untuk baris dengan nilai tersedia.
3. Menandai `Ts_s < Tp_s` sebagai kondisi yang perlu diperiksa.
4. Menampilkan tabel dan ringkasan status.

## CONSTRAINTS
- Jangan mengubah nilai input awal.
- Pertahankan satuan detik.
- Jangan mengisi nilai yang hilang dengan tebakan.
- Jangan menyatakan data valid secara seismologis hanya berdasarkan QC sederhana.
- Nyatakan bahwa data contoh bersifat sintetis.

## OUTPUT FORMAT
Tabel berisi `stasiun`, `Tp_s`, `Ts_s`, `delta_t_s`, dan `status_qc`, diikuti ringkasan jumlah setiap status.

## QUALITY CRITERIA
- Selisih dihitung dengan rumus `Ts_s - Tp_s`.
- Baris dengan input hilang tidak dinyatakan lulus.
- Baris dengan `Ts_s < Tp_s` ditandai.
- Nilai input tetap sama seperti data awal.
- Hasil perhitungan dapat diperiksa ulang.

## VERIFICATION
Hitung manual satu baris, lalu cocokkan dengan hasil kode. Periksa juga bahwa baris dengan nilai hilang atau `Ts_s < Tp_s` ditandai untuk diperiksa.

## RECOMMENDED STACK
Python dan pandas untuk tabel, serta NumPy untuk nilai numerik yang hilang. Stack ini cukup untuk demonstrasi kecil; tidak diperlukan framework tambahan.

## READY-TO-USE PROMPT
> Bertindak sebagai asisten pemrosesan data seismologi. Periksa tabel waktu tiba P dan S yang diberikan. Hitung `delta_t_s = Ts_s - Tp_s` bila kedua nilai tersedia. Tandai nilai yang hilang dan baris dengan `Ts_s < Tp_s`. Jangan mengubah input atau menebak nilai yang hilang. Tampilkan tabel hasil, ringkasan status, asumsi, dan batasan QC. Sebutkan bahwa QC dasar tidak menggantikan pemeriksaan waveform.

## CODE OUTLINE
1. Buat atau baca tabel input.
2. Salin tabel agar input asli tetap terjaga.
3. Hitung selisih S–P.
4. Terapkan aturan status QA/QC.
5. Tampilkan hasil dan ringkasan.

## EXPECTED OUTPUT
Tabel QA/QC dan ringkasan status yang bisa dibandingkan dengan hasil aktual pada cell berikutnya.

## OPTIONAL CHECK
Perhitungan manual untuk `STA01`: `14.70 - 12.40 = 2.30 s`.

## REPRODUCIBILITY NOTE

## COMPANION FILE PATHS
Jika dipublikasikan sebagai project Git, recipe ini dapat disimpan sebagai `recipes/R-GEO-001.md` dan implementasi sebagai `code/R-GEO-001.py`. Path tersebut adalah rancangan struktur, bukan file terpisah yang dibuat oleh notebook ini.

## RELATED RECIPES
Recipe terkait yang masuk akal adalah recipe pembacaan waveform, pemeriksaan metadata stasiun, dan penentuan hiposenter. Recipe tersebut tidak dijalankan dalam latihan ini.

```python
import pandas as pd
import numpy as np

# Data sintetis untuk latihan; bukan data observasi nyata.
data = pd.DataFrame({
    "stasiun": ["STA01", "STA02", "STA03", "STA04", "STA05"],
    "Tp_s": [12.40, 14.10, 11.80, np.nan, 16.20],
    "Ts_s": [14.70, 17.30, 10.90, 19.20, 20.10]
})

# Buat salinan agar data input awal tetap terjaga.
hasil = data.copy()

# Hitung selisih S-P hanya ketika kedua nilai tersedia.
hasil["delta_t_s"] = hasil["Ts_s"] - hasil["Tp_s"]

# Tandai setiap baris berdasarkan pemeriksaan sederhana.
def periksa_baris(baris):
    if pd.isna(baris["Tp_s"]) or pd.isna(baris["Ts_s"]):
        return "PERLU DIPERIKSA: nilai hilang"
    if baris["Ts_s"] < baris["Tp_s"]:
        return "PERLU DIPERIKSA: Ts < Tp"
    return "LULUS QC DASAR"

hasil["status_qc"] = hasil.apply(periksa_baris, axis=1)

display(hasil)
print("\nRingkasan status:")
print(hasil["status_qc"].value_counts())
```

Pemeriksaan ini tidak membuktikan bahwa waktu pick benar secara seismologis. Perlu pemeriksaan waveform, metadata, dan aturan picking yang sesuai sebelum data digunakan untuk analisis ilmiah.

```python
# Optional check: verifikasi manual baris STA01.
tp_manual = 12.40
ts_manual = 14.70
delta_manual = ts_manual - tp_manual

delta_dari_kode = hasil.loc[hasil["stasiun"] == "STA01", "delta_t_s"].iloc[0]

print(f"Hitung manual Δt = {ts_manual:.2f} - {tp_manual:.2f} = {delta_manual:.2f} s")
print(f"Hasil kode untuk STA01 = {delta_dari_kode:.2f} s")
assert np.isclose(delta_manual, delta_dari_kode)
print("CHECK BERHASIL: hasil manual sama dengan hasil kode.")
```
